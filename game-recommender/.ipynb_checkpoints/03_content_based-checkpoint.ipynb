### 3.1 - Import Libraries

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics.pairwise import cosine_similarity
import matplotlib.pyplot as plt
import seaborn as sns
import pickle
import os

print("All libraries imported successfully!")
```

### 3.2 - Load and Explore items.csv

```python
items_df = pd.read_csv('data/items.csv')

print("First 5 rows:")
print(items_df.head())

print(f"\nShape: {items_df.shape}")

print("\nColumn data types:")
print(items_df.dtypes)

print("\nValue counts — category:")
print(items_df['category'].value_counts())

print("\nValue counts — type:")
print(items_df['type'].value_counts())

print("\nValue counts — rarity:")
print(items_df['rarity'].value_counts())
```

### 3.3 - Identify Text vs Numeric Columns

```python
# Columns that need One-Hot Encoding (text/categorical)
text_cols = ['category', 'type', 'rarity']

# Columns that need normalization (numeric stats)
numeric_cols = ['damage', 'range', 'fire_rate', 'price']

# Identifier columns — NOT used as features
id_cols = ['item_id', 'item_name']

print("Text columns (One-Hot Encoding):", text_cols)
print("Numeric columns (MinMax Scaling):", numeric_cols)
print("Identifier columns (excluded from features):", id_cols)
print("\nNote: item_id and item_name are identifiers, not features.")
print("Text columns must be encoded — the model cannot compute similarity on words.")
```

### 3.4 - Apply One-Hot Encoding on Text Columns

```python
items_encoded = pd.get_dummies(items_df, columns=['category', 'type', 'rarity'])

print(f"Shape after One-Hot Encoding: {items_encoded.shape}")
print("\nAll column names after encoding:")
print(list(items_encoded.columns))

print("\nFirst 5 rows (encoded columns shown):")
encoded_cols = [c for c in items_encoded.columns if c.startswith(('category_', 'type_', 'rarity_'))]
print(items_encoded[['item_name'] + encoded_cols[:8]].head(5))
```

### 3.5 - Normalize Numeric Columns with MinMaxScaler

```python
numeric_cols = ['damage', 'range', 'fire_rate', 'price']

scaler = MinMaxScaler()

items_encoded[numeric_cols] = scaler.fit_transform(items_encoded[numeric_cols])

print("Min and Max of each numeric column after scaling:")
print(items_encoded[numeric_cols].agg(['min', 'max']))

print(f"\nShape after scaling (should be unchanged): {items_encoded.shape}")
print("All numeric values are now between 0.0 and 1.0 — no column dominates the similarity score.")
```

### 3.6 - Build the Feature Matrix

```python
feature_matrix = items_encoded.drop(columns=['item_id', 'item_name', 'xp_reward', 'difficulty'], errors='ignore')

print(f"Feature matrix shape: {feature_matrix.shape}")
print("\nFeature columns used:")
print(list(feature_matrix.columns))

missing = feature_matrix.isnull().sum().sum()
print(f"\nMissing values in feature matrix: {missing}")
print("Each row = 1 item. Each column = 1 feature. Ready for cosine similarity.")
```

### 3.7 - Build the Cosine Similarity Matrix

```python
similarity_matrix = cosine_similarity(feature_matrix)

# Convert to DataFrame with item names as index and columns
similarity_matrix = pd.DataFrame(
    similarity_matrix,
    index=items_df['item_name'],
    columns=items_df['item_name']
)

print(f"Similarity matrix shape: {similarity_matrix.shape}")
print("\nPreview (top-left 4×4):")
print(similarity_matrix.iloc[:4, :4].round(3))

print("\nSanity check — Top 3 most similar items to 'Sniper AWM':")
similar_to_awm = similarity_matrix['Sniper AWM'].sort_values(ascending=False)[1:4]
print(similar_to_awm)
```

### 3.8 - Save Similarity Matrix and Feature Matrix

```python
os.makedirs('models', exist_ok=True)

# Save similarity matrix
with open('models/cbf_similarity.pkl', 'wb') as f:
    pickle.dump(similarity_matrix, f)

# Save feature matrix
with open('models/cbf_features.pkl', 'wb') as f:
    pickle.dump(feature_matrix, f)

print("Files saved:")
print(f"  models/cbf_similarity.pkl exists: {os.path.exists('models/cbf_similarity.pkl')}")
print(f"  models/cbf_features.pkl exists: {os.path.exists('models/cbf_features.pkl')}")

sim_size = os.path.getsize('models/cbf_similarity.pkl') / 1024
feat_size = os.path.getsize('models/cbf_features.pkl') / 1024
print(f"  cbf_similarity.pkl size: {sim_size:.2f} KB")
print(f"  cbf_features.pkl size:   {feat_size:.2f} KB")
```

### 3.9 - Build the Recommendation Function

```python
interactions_df = pd.read_csv('data/interactions.csv')

def get_cbf_recommendations(player_id, n=5):
    """
    Returns top-N content-based recommendations for a player.
    Bases recommendations on items the player rated 4 or 5 stars.
    Falls back to highest-rated item if none rated 4+.
    Excludes items the player has already rated.
    """
    player_data = interactions_df[interactions_df['player_id'] == player_id]

    # Items rated 4+
    liked_items = player_data[player_data['rating'] >= 4]['item_name'].tolist()

    # Fallback: use highest-rated item if no 4+ ratings
    if not liked_items:
        best_row = player_data.sort_values('rating', ascending=False).iloc[0]
        liked_items = [best_row['item_name']]

    # All items this player has already rated (exclude from recommendations)
    already_rated = set(player_data['item_name'].tolist())

    # Aggregate similarity scores across all liked items
    combined_scores = pd.Series(0.0, index=similarity_matrix.columns)
    for item in liked_items:
        if item in similarity_matrix.index:
            combined_scores += similarity_matrix[item]

    # Average across liked items
    combined_scores = combined_scores / len(liked_items)

    # Remove already-rated items
    combined_scores = combined_scores.drop(labels=list(already_rated), errors='ignore')

    # Sort and return top N
    top_n = combined_scores.sort_values(ascending=False).head(n)
    return list(zip(top_n.index, top_n.values))

print("Recommendation function defined successfully!")
print(f"Interactions loaded: {interactions_df.shape[0]} rows, {interactions_df['player_id'].nunique()} unique players")
```

### 3.10 - Test the Function on 3 Players

```python
players_df = pd.read_csv('data/players.csv')

test_players = [1, 25, 50]

for pid in test_players:
    play_style = players_df[players_df['player_id'] == pid]['play_style'].values[0]
    liked = interactions_df[
        (interactions_df['player_id'] == pid) & (interactions_df['rating'] >= 4)
    ]['item_name'].tolist()

    recs = get_cbf_recommendations(pid, n=5)

    print(f"{'='*75}")
    print(f"Player {pid} | Play style: {play_style}")
    print(f"Liked items (rated 4+): {liked if liked else 'None (using fallback)'}")
    print(f"Top 5 Recommendations:")
    for rank, (item, score) in enumerate(recs, 1):
        print(f"  {rank}. {item:<30} (similarity: {score:.4f})")
    print()
```

###  3.11 - Similarity Heatmap

```python
os.makedirs('outputs', exist_ok=True)

fig, ax = plt.subplots(figsize=(18, 14))

sns.heatmap(
    similarity_matrix,
    cmap='YlOrRd',
    vmin=0,
    vmax=1,
    xticklabels=True,
    yticklabels=True,
    ax=ax
)

ax.set_title('Item-Item Cosine Similarity Matrix (Content-Based Filtering)', fontsize=14, pad=15)
ax.tick_params(axis='x', labelsize=7)
ax.tick_params(axis='y', labelsize=7)
plt.xticks(rotation=90)
plt.yticks(rotation=0)
plt.tight_layout()

plt.savefig('outputs/similarity_heatmap.png', dpi=150)
plt.show()
print("Heatmap saved to outputs/similarity_heatmap.png")
print("Look for: bright red diagonal (self-similarity = 1.0) and red clusters by item type.")
```

### 3.12 - Generate Recommendations for All 100 Players

```python
all_cbf_results = []

all_player_ids = interactions_df['player_id'].unique()

for pid in all_player_ids:
    recs = get_cbf_recommendations(pid, n=5)
    for rank, (item_name, score) in enumerate(recs, 1):
        all_cbf_results.append({
            'player_id': pid,
            'item_name': item_name,
            'similarity_score': round(score, 4),
            'rank': rank
        })

cbf_results_df = pd.DataFrame(all_cbf_results)
cbf_results_df.to_csv('outputs/cbf_results.csv', index=False)

print(f"Total rows saved: {len(cbf_results_df)}")
print(f"Players covered: {cbf_results_df['player_id'].nunique()}")
print("\nFirst 10 rows:")
print(cbf_results_df.head(10).to_string(index=False))
```

### 3.13 - Summary

```python
print("=" * 75)
print("CONTENT-BASED FILTERING — SUMMARY")
print("=" * 75)
print(f"  Total items in similarity matrix : {similarity_matrix.shape[0]}")
print(f"  Total players covered            : {cbf_results_df['player_id'].nunique()}")
print(f"  Total recommendations generated  : {len(cbf_results_df)}")

print("\n--- Top 5 Most Similar Item Pairs (excluding self) ---")
sim_copy = similarity_matrix.copy()
np.fill_diagonal(sim_copy.values, 0)
stacked = sim_copy.stack()
top_pairs = stacked.sort_values(ascending=False).head(10)
seen = set()
count = 0
for (item1, item2), score in top_pairs.items():
    pair = frozenset([item1, item2])
    if pair not in seen:
        seen.add(pair)
        print(f"  {count+1}. {item1} <-> {item2}: {score:.4f}")
        count += 1
        if count == 5:
            break

print("\n--- Output Files Created ---")
files = [
    'models/cbf_similarity.pkl',
    'models/cbf_features.pkl',
    'outputs/similarity_heatmap.png',
    'outputs/cbf_results.csv'
]
for f in files:
    exists = os.path.exists(f)
    size = os.path.getsize(f) / 1024 if exists else 0
    print(f"  {'✓' if exists else '✗'} {f} ({size:.1f} KB)")

print("\n--- Sample Recommendation for Player 1 ---")
sample_recs = get_cbf_recommendations(1, n=5)
for rank, (item_name, score) in enumerate(sample_recs, 1):
    category = items_df[items_df['item_name'] == item_name]['category'].values
    cat_label = category[0] if len(category) > 0 else 'Unknown'
    print(f"  {rank}. {item_name:<30} | Category: {cat_label:<10} | Score: {score:.4f}")

print("\n✓ Notebook complete.")
```
